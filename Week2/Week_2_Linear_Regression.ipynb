# Week 2 Project — Linear Regression
## Diabetes Progression Prediction

This project uses Pandas for data handling and Scikit-learn Linear Regression to predict a target value.

**Week 2 concepts:** train/test split, linear regression, MAE, MSE, RMSE and R².


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 1. Load the dataset


```python
data = load_diabetes(as_frame=True)
df = data.frame

print("Dataset shape:", df.shape)
df.head()
```

## 2. Understand and clean the data


```python
print(df.info())
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()
df.describe()
```

The **target** column is the value we want to predict. The remaining columns are the input **features**.


```python
X = df.drop(columns=["target"])
y = df["target"]

print("Features:", X.shape)
print("Target:", y.shape)
```

## 3. Train/Test Split
80% of the data is used for training and 20% for testing.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

## 4. Train the Linear Regression model


```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Model trained successfully!")
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

## 5. Make predictions


```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})
results.head(10)
```

## 6. Evaluate the model


```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.2f}")
```

### What the metrics mean
- **MAE:** average absolute prediction error. Lower is better.
- **MSE:** average squared prediction error. Lower is better.
- **RMSE:** error in the same units as the target. Lower is better.
- **R²:** how much variation in the target is explained by the model. Higher is better.


## 7. Visualize Actual vs Predicted values


```python
plt.figure(figsize=(7,5))
plt.scatter(y_test, y_pred)
plt.xlabel("Actual values")
plt.ylabel("Predicted values")
plt.title("Actual vs Predicted — Linear Regression")
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()])
plt.show()
```

## 8. Conclusion
A Linear Regression model was trained using an 80/20 train-test split. The model was evaluated using MAE, MSE, RMSE and R². The project demonstrates the complete basic machine-learning workflow: load data → inspect/clean → select features and target → split data → train model → predict → evaluate.
